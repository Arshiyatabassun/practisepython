```python
#Data structures:list,tuple,dic,sets
# List:
# it has collection of values
# it allows duplicate values
# it is mutable data type
# it contains combination of different data types
# it stores any type of data
# index postion always starts from 0
# list is enclosed with []
```

```python
list =[10,20,30,"hi",5.5,"hey",50,40,20,20]
print(list[1])
```

```python
#Methods in list:
# update:
list[0]=100
list[4]="arshi"
list
```

```python
# insert:we can update the value using index position
list.insert(0,"mohisn")
list
```

```python
# append:we can add only one value at the last index position
list1=[11,12,13,14,15,16,17]
list1.append(200)
list1
```

```python
#extend:we can add morethan one value at the last index
list1.extend([100,200,300,400])
list1
```

```python
#del:we can delete the data using last index position deletes the structure also and del returns nothing
del list1[4]
del list1[0:3]
del list1
list1
```

```python

# clear:it deletets the entire data from the list bt structure remains constant
list2=['hey',12,13,42,44,56,"mmom"]
list2.clear()
list2
```

```python
# pop:by default it remove the last index postiom
list3=[20,30,40,50,67]
list3.pop(3)
list3
```

```python
# remove :we cam delete the value directly
list3.remove(40)
list3
```

```python
# reverse:we can reversse an entire list
list4=[10,20,30,40,50,60,70]
list4.reverse()
list4
```

```python
# sort:arrange the values form lowest ot highest
list4.sort()
list4
```

```python
# sortboolean:to print the values in reverse order
list4.sort(reverse=True)
list4
```

```python
# len:
len(list4)
```

```python
# count:
list4.count(50)
```

```python
# min,max
min(list4)
```

```python
max(list4)
```
